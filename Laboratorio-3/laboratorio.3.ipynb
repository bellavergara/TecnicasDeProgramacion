• si el ingreso del ciudadano no era superior a $85528, el impuesto era igual al 18% del ingreso

• si el ingreso era superior a esta cantidad, el impuesto era igual a $14839.02, más el 32% del

Información Privada

• Si el impuesto calculado es menor que cero, solo significa que no hay impuesto (el impuesto es
igual a cero).

A continuación, debe imprimir el impuesto calculado, redondeado a pesos totales. Hay una función
llamada round() que hará el redondeo

Como seguramente sabrás, debido a algunas razones astronómicas, el año puede ser bisiesto o
común. Los primeros tienen una duración de 366 días, mientras que los últimos tienen una duración
de 365 días.
Desde la introducción del calendario Gregoriano (en 1582), se utiliza la siguiente regla para determinar
el tipo de año:
• si el número del año no es divisible entre cuatro, es un año común.
• de lo contrario, si el número del año no es divisible entre 100, es un año bisiesto.
• de lo contrario, si el número del año no es divisible entre 400, es un año común.
• de lo contrario, es un año bisiesto.
Observa el código en el editor - solo lee un número de año y debe completarse con las instrucciones
que implementan la prueba que acabamos de describir.
El código debe mostrar uno de los dos mensajes posibles, que son Año Bisiesto o Año Común, según
el valor ingresado.
Sería bueno verificar si el año ingresado cae en la era Gregoriana y emitir una advertencia de lo
contrario: No dentro del período del calendario Gregoriano. Consejo: utiliza los operadores != y %.

```python
year = int(input("introduce un año:"))
if year <1582:
    print("no esta dentro de; calendario gregoriano")
else:
    if year % 4 !=0:
        print(" año comun")
    elif year % 100 !=0:
        print("año bisiesto")
    elif year % 400 !=0:
        print("año comun")
    else:
        print("año bisiesto")




```
